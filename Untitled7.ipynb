```python
num=df.select_dtypes('number')
(num.describe(include='all'))
```

```python
rows = []
for col in num.columns:
    q1, q3 = num[col].quantile([.25, .75])
    iqr = q3 - q1
    low, high = q1 - 1.5*iqr, q3 + 1.5*iqr
    n = ((num[col] < low) | (num[col] > high)).sum()
    rows.append((col, num[col].min(), num[col].max(), n))

print(pd.DataFrame(rows, columns=["column", "min", "max", "iqr_outliers"]))
```

```python
import matplotlib.pyplot as plt

fig, axes = plt.subplots(2, 2, figsize=(11, 8))

# 1. Churn rate by contract
rate = df.groupby('Contract')['Churn Value'].mean().mul(100).round(1)
rate.plot(kind='bar', color='steelblue', ax=axes[0, 0])
axes[0, 0].bar_label(axes[0, 0].containers[0], fmt='%.1f%%')
axes[0, 0].set_title('Churn rate by contract type')
axes[0, 0].set_ylabel('Churn rate (%)')
axes[0, 0].tick_params(axis='x', rotation=0)

# 2. Churn rate by tenure group
rate = df.groupby('Tenure Group', observed=True)['Churn Value'].mean().mul(100).round(1)
rate.plot(kind='bar', color='steelblue', ax=axes[0, 1])
axes[0, 1].bar_label(axes[0, 1].containers[0], fmt='%.1f%%')
axes[0, 1].set_title('Churn rate by tenure (months)')
axes[0, 1].set_ylabel('Churn rate (%)')
axes[0, 1].tick_params(axis='x', rotation=0)

# 3. Reasons given by churned customers
reasons = df[df['Churn Value'] == 1]['Churn Category'].value_counts(normalize=True).mul(100).round(1)
reasons.sort_values().plot(kind='barh', color='steelblue', ax=axes[1, 0])
axes[1, 0].bar_label(axes[1, 0].containers[0], fmt='%.1f%%')
axes[1, 0].set_title('Reasons given by churned customers')
axes[1, 0].set_xlabel('Share of churned customers (%)')

# 4. Top cities by churn rate
city = df.groupby('City')['Churn Value'].agg(['count', 'mean'])
top = city[city['count'] >= 30].sort_values('mean', ascending=False).head(10)['mean'].mul(100).round(1)
top.sort_values().plot(kind='barh', color='steelblue', ax=axes[1, 1])
axes[1, 1].bar_label(axes[1, 1].containers[0], fmt='%.1f%%')
axes[1, 1].set_title('Top 10 cities by churn rate (30+ customers)')
axes[1, 1].set_xlabel('Churn rate (%)')

plt.tight_layout()
fig.savefig('churn_dashboard.png', dpi=150, bbox_inches='tight', facecolor='white')
plt.show()
```
